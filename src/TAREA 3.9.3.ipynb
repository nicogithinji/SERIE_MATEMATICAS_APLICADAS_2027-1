```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
c = 1
t = [0, 1, 2, 3, 4]
```

```python
def h(x,t):
    if 0 <= t - x/c <= 2:
        return 0.5 * (t - x/c)
    elif 2 <= t - x/c <= 3:
        return 3 - t - x / c
    elif t - x / c < 0:
        return 0
```

```python
x = np.linspace(0,5 * c, 100)
y = np.zeros_like(x)
```

```python
for tiempo in range(len(t)):
    for n in range(len(x)):
        y[n] = h(x[n], t[tiempo])
    plt.plot(x,y, label =f" $t = {t[tiempo]}$")

plt.xlabel("$x$")
plt.ylabel(r"$u(x,t)$")
plt.tight_layout()
plt.grid(True)
plt.legend()
```
