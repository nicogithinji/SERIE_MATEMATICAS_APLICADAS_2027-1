```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import math
from pandas import DataFrame, Series
```

```python
a = 1
h = 1
c = 1
t = [0, a / (6 * c), a / (2 * c), 5 * a / (6 * c)]
t_label = ["0", "a/6c", "a/2c", "5a/6c"]
```

```python
def F(x):
    if a < x or -a > x:
        return 0
    if 0 < x < (2 * a / 3):
        return (3 * h * x) / (2 * a)
    if (2 * a / 3) < x < a:
        return (3 * h * (a - x) / a)
    if (-2 * a / 3) < x < 0:
        return (3 * h * x / (2 * a))
    if (-2 * a / 3) > x > (-a):
        return (- 3 * h * (a + x))
    else:
        return 0
```

```python
x = np.linspace(-a-1, a+1)
y = np.zeros_like(x)
```

```python
for tiempo in range(len(t)):
    for n in range(len(x)):
        y[n] = 0.5 * (F(x[n] + c * t[tiempo]) + F(x[n] - c * t[tiempo]))
    plt.plot(x,y, label =f" $t = {t_label[tiempo]}$")
plt.xlabel("$x$")
plt.ylabel(r"$u(x,t)$")
plt.tight_layout()
plt.grid(True)
plt.legend()
```
