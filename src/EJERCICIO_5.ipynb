```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
c = 1
t = [0, 1/(2*c), 1/c, 2/c, 3/c]
t_label = ["0", "1/2c", "1/c", "2/c", "3/c"]
```

```python
def F(x):
    if x <= -2 or x >= 2:
        return 0
    elif -2 < x < -1:
        return -1
    elif -1 <= x <= 1:
        return 0
    elif 1 < x < 2:
        return 1
    else:
        return 0
```

```python
x = np.linspace(0, 6)
y = np.zeros_like(x)
```

```python
for tiempo in range(len(t)):
    for n in range(len(x)):
        y[n] = 0.5 * (F(x[n] + c * t[tiempo]) + F(x[n] - c * t[tiempo]))
    plt.plot(x,y, label =f" $t = {t_label[tiempo]}$")

plt.xlabel("$x$")
plt.ylabel(r"$u(x,t)$")
plt.tight_layout()
plt.grid(True)
plt.legend()
```
