```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
c = 1
t = [0, np.pi/2, np.pi, 3 * np.pi / 2, 5 * np.pi / 2]
t_label = [r"0", r"\pi/2", r"\pi", r"3\pi/2", r"5\pi/2"]
```

```python
def h(x,t):
    if t >= x / c:
        return np.sin(t - x / c)
    else:
        return 0
```

```python
x = np.linspace(0,10,100)
y = np.zeros_like(x)
```

```python
for tiempo in range(len(t)):
    for n in range(len(x)):
        y[n] = h(x[n], t[tiempo])
    plt.plot(x,y, label =f" $t = {t_label[tiempo]}$")

plt.xlabel("$x$")
plt.ylabel(r"$u(x,t)$")
plt.tight_layout()
plt.grid(True)
plt.legend()
```
